```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the correct dataset from the workspace
df = pd.read_csv("/content/student_performance_dataset.csv")
print("Dataset Shape : ", df.shape)
print(df.head())
print("Dataset Information : ")
df.info()
print("Dataset Description : \n", df.describe())

# Distribution of Final Exam Score
plt.figure(figsize=(6,4))
sns.histplot(df["Final_Exam_Score"], bins=20, kde=True, color="steelblue")
plt.title("Distribution of Final Exam Score")
plt.xlabel("Final Exam Score")
plt.ylabel("Frequency")
plt.show()

# Final Exam Score across Study Hours per Week
plt.figure(figsize=(10,5))
sns.boxplot(x="Study_Hours_per_Week", y="Final_Exam_Score", data=df, color="steelblue")
plt.title("Final Exam Score across Study Hours per Week")
plt.xlabel("Study Hours per Week")
plt.ylabel("Final Exam Score")
plt.xticks(rotation=45)
plt.show()

# Past Exam Scores vs Final Exam Score
plt.figure(figsize=(6,4))
sns.scatterplot(x="Past_Exam_Scores", y="Final_Exam_Score", hue="Gender", data=df, palette="coolwarm")
plt.title("Past Exam Scores vs Final Exam Score")
plt.xlabel("Past Exam Scores")
plt.ylabel("Final Exam Score")
plt.show()

# Pairwise relationships for selected numerical variables
sns.pairplot(df[["Study_Hours_per_Week", "Attendance_Rate", "Past_Exam_Scores", "Final_Exam_Score", "Gender"]], hue="Gender", palette="coolwarm")
plt.suptitle("Pair Relations", y=1.02)
plt.show()

# Correlation Heatmap for numerical variables
numeric_cols = ["Study_Hours_per_Week", "Attendance_Rate", "Past_Exam_Scores", "Final_Exam_Score"]
plt.figure(figsize=(8,6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
